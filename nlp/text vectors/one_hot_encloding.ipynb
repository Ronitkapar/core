```python
# Implementadtion

import pandas as pd
df = pd.read_csv('nlp_practice_dataset.csv',usecols=['sentence'])
```

```python
df.head(15)
```

```python
# Data CLeaning and preprocessing
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import RegexpStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer

ps= PorterStemmer()
rs = RegexpStemmer('ing$|s$|e$|able$|ed$|ly$|ment$|ness$|ity$|al$|er$|ous$|ive$|ful$|less$')
snowball = SnowballStemmer('english')
lemitizer = WordNetLemmatizer()
stopwords= set(stopwords.words('english'))
```

```python
# Retriving sentences from the datafrane and string it in a list for better processing. 
sentences = []
for i in df['sentence']:
    sentences.append(i)

print(f"The original sentence is: {sentences}")
```

```python
words=[]
for i in sentences:
    #Spliting each sentences into words
    words.extend(i.split())
#words =[word.lower() for word in words ]   
print(words)
```

```python
# Tokenize the sentence into words via different methods.

ps_words = [ps.stem(word) for word in words]
print(ps_words)
```

```python
rs_words = [rs.stem(word) for word in words]
print(rs_words)
```

```python
snowball_words = [snowball.stem(word) for word in words]
print(snowball_words)
```

```python
lemitizer_words = [lemitizer.lemmatize(word.lower(), pos='v') for word in words]
print(lemitizer_words)
```

```python
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

cleaned_sentences = []

for sentence in sentences:
    words = word_tokenize(sentence)
    
    filtered = [
        word for word in words
        if word.lower() not in stop_words
    ]
    
    cleaned_sentences.append(" ".join(filtered))

print(cleaned_sentences)
```

```python
## Above code is breaking the sentences into words, we cannt encode the sentences into vectors. We need to keep the sentences as it is and remove the stop words from the sentences.
## so the new approach is given below.
```

```python
from nltk.tokenize import word_tokenize
stemmed_sentences = []
for sentence in sentences:
    
    #Tokenize the sentence into words
    words = word_tokenize(sentence)
    
    #applied PotterStemm to each word in the sentence and convert it into lower case
    stemmed_words = [ps.stem(word.lower()) for word in words]

    #joining each words back to form sentences
    stemmed_sentences.append(" ".join(stemmed_words))
   
print(stemmed_sentences)

```

```python
# with stop words removal

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

stemmed_sentences1 = []

for sentence in sentences:
    
    # tokenize the sentence into words
    words= word_tokenize(sentence)

    # remove stop words and apply PorterStemmer to each word in the sentence and convert it into lower case
    processed = [
        ps.stem(word.lower()) 
        for word in words
        if word.lower() not in stop_words
    ]

    stemmed_sentences1.append(" ".join(processed))

print(stemmed_sentences1)
```

```python
# Trying with snowman

from nltk.stem import SnowballStemmer
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

#stop_words = set(stopwords.words('english'))
snowball = SnowballStemmer('english')

snowball_sentences = []

for sentence in sentences:
    
    #Tokenize the sentences into words
    words = word_tokenize(sentence)

    #remove stop words and apply snowball stemmer to each word in the sentence and convert it into lower case
    filtered_words = [
        snowball.stem(word.lower())
        for word in words
        #if word.lower() not in stop_words
        if word.lower() not in stopwords.words('english')

        
    ]
    snowball_sentences.append(" ".join(filtered_words))

print(snowball_sentences)

```

```python
# trying with lemmatization

import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemitizer = WordNetLemmatizer()
lemitized_sentences = []

for sentence in sentences:

    words = word_tokenize(sentence)

    lemittized_words = [
        lemitizer.lemmatize(word, pos = 'v')
        for word in words 
        if word not in stopwords.words('english')
    ]

    lemitized_sentences.append(" ".join(lemittized_words))

print(lemitized_sentences)
```

```python
with open('cleaned_sentences.txt', 'a') as f:
    for sentence in lemitized_sentences:
        f.write(sentence + '\n')
```
