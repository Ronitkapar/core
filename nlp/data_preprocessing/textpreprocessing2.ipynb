# Text preporcessing using different method

## 1. Stemming

```python
import nltk
paragraph = 'Five score years ago a great American in whose symbolic shadow we stand today signed the Emancipation Proclamation. This momentous decree is a great beacon light of hope to millions of Negro slaves who had been seared in the flames of withering injustice. It came as a joyous daybreak to end the long night of their captivity. But 100 years later the Negro still is not free. One hundred years later the life of the Negro is still badly crippled by the manacles of segregation and the chains of discrimination. One hundred years later the Negro lives on a lonely island of poverty in the midst of a vast ocean of material prosperity. One hundred years later the Negro is still languished in the corners of American society and finds himself in exile in his own land. So we’ve come here today to dramatize a shameful condition.'
```

```python
from nltk import word_tokenize
words = word_tokenize(paragraph)
print(words)
```

```python
# Steaming is a wat to reduce a word to its root stem. It affect affixes to suffies and prefixes. The roots of the words is knows as a lemma.

import nltk 
from nltk.stem import PorterStemmer  #PorterStemmer is the keyword to import this specific function

stemming = PorterStemmer()
for w in words:
    print(f"{w} ----> {stemming.stem(w)}")
```

```python
# Here in the above code. I did the following thing:
# Took a paragraph --> tokenized to word --> stemmed: using for loop, took each word one by one from words.
```

## RegexStemmer
####  It is highly customizable stemming technique used in Natural Language Processing (NLP) that strips morphological affixes (prefixes or suffixes) from words using regular expressions

```python
from nltk.stem import RegexpStemmer
reg_stemming = RegexpStemmer('ing$|s$|e$|able$', min=3)  # The 1st parameter is the regex pattern for matching suffixes to remove.

for word in words:
    print(f"{word} ----> {reg_stemming.stem(word)}")
```

## SnowBall Stemmer
#### a fast, rule-based algorithm designed by Martin Porter to reduce words to their base forms (stems) across multiple languages

```python
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english')

for word in words:
    print(f"{word} ----> {snowball.stem(word)}")
```

## Lemmatizer 
#### It converts words to their canonical dictionary root (lemma). It evaluates morphological structures using the extensive WordNet lexical database, which guarantees the output is an actual, meaningful word.

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')  # Have to downlaod wordnet brfore using it as it compares the word with the dictionary and gives the correct lemma.

lemmatizer = WordNetLemmatizer()
for word in words:
    print(f"{word} ----> {lemmatizer.lemmatize(word, pos = 'n')}")  # pos is used to specify which part of speech the word is. [By defult it is noun].
```

## StopWords
#### highly frequent, low-value words (like "the," "is," "and," "in") that lack semantic meaning. Their removal during text preprocessing reduces dataset size, accelerates computation, and helps algorithms focus on context-rich

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english')) #stop words contain words -> common words which needs to be removed. for eg: 'this', 'is', 'and'...
print(f"stop_words: {stop_words}")

filtered_words = [word for word in words if word not in stop_words] # created a new list where stop words are removed. for eg: 'this', 'is', 'and'... are removed from the list of words.
print(f"filtered words = {filtered_words}")

cleaned_words = " ".join(filtered_words) # joined the words for cleaner look. 
print(f"Cleaned words: {cleaned_words}")
```

## Parts of Speech
#### find parts of speech of each word. important for lemmitization

```python
import nltk
from nltk import word_tokenize, pos_tag

nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt')

paragraph = 'Five score years ago a great American in whose symbolic shadow we stand today signed the Emancipation Proclamation. This momentous decree is a great beacon light of hope'
paragraph = paragraph.lower()

words = word_tokenize(paragraph)
print(words)

tagged_words = pos_tag(words)

for word, tag in tagged_words:
    print(f"{word} ----> {tag}")
```
